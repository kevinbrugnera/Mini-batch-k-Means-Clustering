```python
import os
import time
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_rcv1
from sklearn.decomposition import TruncatedSVD

rcv1 = fetch_rcv1()

max_components = 500

    
start_time = time.time()
    
# Fit the SVD once with the maximum desired components
svd = TruncatedSVD(n_components=max_components, random_state=42)
svd.fit(rcv1.data)

# Calculate the cumulative sum of the explained variance percentages
cumulative_variance = np.cumsum(svd.explained_variance_ratio_) * 100
    
    
# Create the plot figure
plt.figure(figsize=(10, 6))
plt.plot(range(1, max_components + 1), cumulative_variance, linewidth=2.5, color='#1f77b4')
    
# Formatting the plot for a professional presentation
plt.title("Scree Plot: Cumulative Explained Variance vs. Number of Components", fontsize=14, fontweight='bold')
plt.xlabel("Number of Latent Components", fontsize=12)
plt.ylabel("Cumulative Explained Variance (%)", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
    
#Highlight specific milestones (e.g., 100 components)
plt.axvline(x=100, color='r', linestyle=':', label='100 Components')
plt.legend(loc='lower right', fontsize=10)
    
# Save the plot in high resolution
output_filename = "explained_variance_plot.png"
plt.savefig(output_filename, dpi=300, bbox_inches='tight')

    
# Display the plot on screen
plt.show()
```
